```python
import pandas as pd
import numpy as np
import os
```

```python
state = pd.read_csv('data//state.csv')
state.head()
```

```python
def clean_names(df):
    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
        .str.replace(r'[^\w\s]', '', regex=True)
        .str.replace(r'[\s-]+', '_', regex=True)
    )
    return df
```

```python
state = clean_names(state)
print(state.columns)
state[['population']].mean()
```
